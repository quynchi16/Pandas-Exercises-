### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv).

### Step 3. Assign it to a variable called chipo.

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv'
chipo = pd.read_csv(url, sep = '\t')
display(chipo.head())
```

### Step 4. How many products cost more than $10.00?

```python
chipo['item_price'] = chipo['item_price'].str.replace('$', '', regex=False).astype(float)

products_over_10 = chipo[chipo['item_price'] > 10.00]

num_unique_products_over_10 = products_over_10['item_name'].nunique()

print(f"Number of products that cost more than $10.00: {num_unique_products_over_10}")
```

### Step 5. What is the price of each item?
###### print a data frame with only two columns item_name and item_price

```python
item_prices = chipo[['item_name', 'item_price']].drop_duplicates()
display(item_prices)
```

### Step 6. Sort by the name of the item

```python
sorted_item_prices = item_prices.sort_values(by='item_name')
display(sorted_item_prices)
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
chipo['price_per_item'] = chipo['item_price'] / chipo['quantity']

most_expensive_item_idx = chipo['price_per_item'].idxmax()

most_expensive_item_row = chipo.loc[most_expensive_item_idx]

quantity_most_expensive = most_expensive_item_row['quantity']

print(f"The quantity of the most expensive item ordered is: {quantity_most_expensive}")
print(f"Details of the most expensive item:\n{most_expensive_item_row[['item_name', 'price_per_item', 'quantity']]}")
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_bowl_orders = chipo[chipo['item_name'] == 'Veggie Salad Bowl']
num_veggie_salad_bowl_orders = veggie_salad_bowl_orders.shape[0]

print(f"A 'Veggie Salad Bowl' was ordered {num_veggie_salad_bowl_orders} times.")
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
canned_soda_orders = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]
num_canned_soda_multiple = canned_soda_orders.shape[0]

print(f"Someone ordered more than one 'Canned Soda' {num_canned_soda_multiple} times.")
```
