### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/wind.data)

### Step 3. Assign it to a variable called data and replace the first 3 columns by a proper datetime index.

```python
data = pd.read_csv('/content/wind_stats.tsv', sep=r'\s+')

data['Date'] = pd.to_datetime(data[['Yr', 'Mo', 'Dy']].astype(str).agg('-'.join, axis=1))

def fix_year(date):
    if date.year > 2000:
        return date.replace(year=date.year - 100)
    return date

data['Date'] = data['Date'].apply(fix_year)

data = data.set_index('Date')

data.drop(columns=['Yr', 'Mo', 'Dy'], inplace=True)

print(data.info())
print(data.head())
```

### Step 4. Year 2061? Do we really have data from this year? Create a function to fix it and apply it.

```python
def fix_year(date):
    if date.year > 2000:
        return date.replace(year=date.year - 100)
    return date

data.index = data.index.map(fix_year)

print("Data after fixing years (head):")
display(data.head())

print("\nData after fixing years (tail):")
display(data.tail())
```

### Step 5. Set the right dates as the index. Pay attention at the data type, it should be datetime64[ns].

```python
print("Data Index Type:", data.index.dtype)
print("\nDataFrame Info:")
data.info()
```

### Step 6. Compute how many values are missing for each location over the entire record.  
#### They should be ignored in all calculations below.

```python

missing_values_per_location = data.isnull().sum()
print("Missing values per location:")
print(missing_values_per_location)
```

### Step 7. Compute how many non-missing values there are in total.

```python
total_non_missing_values = data.notnull().sum().sum()
print("Total non-missing values:", total_non_missing_values)
```

### Step 8. Calculate the mean windspeeds of the windspeeds over all the locations and all the times.
#### A single number for the entire dataset.

```python
mean_windspeeds_overall = data.stack().mean()
print("Mean windspeeds over all locations and all times:", mean_windspeeds_overall)
```

### Step 9. Create a DataFrame called loc_stats and calculate the min, max and mean windspeeds and standard deviations of the windspeeds at each location over all the days

#### A different set of numbers for each location.

```python
loc_stats = data.agg(['min', 'max', 'mean', 'std'])

print("Location Statistics (loc_stats):")
print(loc_stats)
```

### Step 10. Create a DataFrame called day_stats and calculate the min, max and mean windspeed and standard deviations of the windspeeds across all the locations at each day.

#### A different set of numbers for each day.

```python
day_stats = pd.DataFrame({
    'min': data.min(axis=1),
    'max': data.max(axis=1),
    'mean': data.mean(axis=1),
    'std': data.std(axis=1)
})

print("Daily Statistics (day_stats):")
print(day_stats.head())
```

### Step 11. Find the average windspeed in January for each location.  
#### Treat January 1961 and January 1962 both as January.

```python
january_windspeeds = data[data.index.month == 1]
average_january_windspeeds = january_windspeeds.mean()

print("Average windspeed in January for each location:")
print(average_january_windspeeds)
```

### Step 12. Downsample the record to a yearly frequency for each location.

```python
yearly_data = data.resample('A').mean()

print("Yearly frequency data (first 5 rows):")
print(yearly_data.head())
```

### Step 13. Downsample the record to a monthly frequency for each location.

```python
monthly_data = data.resample('M').mean()

print("Monthly frequency data (first 5 rows):")
print(monthly_data.head())
```

### Step 14. Downsample the record to a weekly frequency for each location.

```python
weekly_data = data.resample('W-MON').mean()

print("Weekly frequency data (first 5 rows):")
print(weekly_data.head())
```

### Step 15. Calculate the min, max and mean windspeeds and standard deviations of the windspeeds across all locations for each week (assume that the first week starts on January 2 1961) for the first 52 weeks.

```python
weekly_stats = pd.DataFrame({
    'min': weekly_data.min(axis=1),
    'max': weekly_data.max(axis=1),
    'mean': weekly_data.mean(axis=1),
    'std': weekly_data.std(axis=1)
})

first_52_weeks_stats = weekly_stats.head(52)

print("Statistics for the first 52 weeks (min, max, mean, std across locations per week):")
print(first_52_weeks_stats)
```
