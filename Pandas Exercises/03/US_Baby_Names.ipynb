# US - Baby Names

### Introduction:

We are going to use a subset of [US Baby Names](https://www.kaggle.com/kaggle/us-baby-names) from Kaggle.  
In the file it will be names from 2004 until 2014


### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/US_Baby_Names_right.csv).

### Step 3. Assign it to a variable called baby_names.

```python
baby_names = pd.read_csv('https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/US_Baby_Names_right.csv')
```

### Step 4. See the first 10 entries

```python
display(baby_names.head(10))
```

### Step 5. Delete the column 'Unnamed: 0' and 'Id'

```python
baby_names.drop(columns=['Unnamed: 0', 'Id'], inplace=True)
```

### Step 6. Is there more male or female names in the dataset?

```python
gender_counts = baby_names['Gender'].value_counts()
display(gender_counts)
```

### Step 7. Group the dataset by name and assign to names

```python
names = baby_names.groupby('Name')['Count'].sum()
display(names.head())
```

### Step 8. How many different names exist in the dataset?

```python
num_unique_names = names.count()
print(f"Number of different names: {num_unique_names}")
```

### Step 9. What is the name with most occurrences?

```python
name_most_occurrences = names.idxmax()
count_most_occurrences = names.max()
print(f"The name with most occurrences is '{name_most_occurrences}' with {count_most_occurrences} occurrences.")
```

### Step 10. How many different names have the least occurrences?

```python
least_occurrences_count = names.min()
names_least_occurrences = names[names == least_occurrences_count]
print(f"Names with the least occurrences ({least_occurrences_count}):")
display(names_least_occurrences)
```

### Step 11. What is the median name occurrence?

```python
median_occurrence = names.median()
print(f"Median name occurrence: {median_occurrence}")
```

### Step 12. What is the standard deviation of names?

```python
std_occurrence = names.std()
print(f"Standard deviation of name occurrences: {std_occurrence}")
```

### Step 13. Get a summary with the mean, min, max, std and quartiles.

```python
summary_stats = names.describe()
display(summary_stats)
```
