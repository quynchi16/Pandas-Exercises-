### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://github.com/thieu1995/csv-files/blob/main/data/pandas/drinks.csv).

### Step 3. Assign it to a variable called drinks.

```python
drinks = pd.read_csv('/content/drinks.csv')
display(drinks.head())
```

### Step 4. Which continent drinks more beer on average?

```python
continent_beer_average = drinks.groupby('continent')['beer_servings'].mean()
display(continent_beer_average)
```

### Step 5. For each continent print the statistics for wine consumption.

```python
wine_stats_by_continent = drinks.groupby('continent')['wine_servings'].describe()
display(wine_stats_by_continent)
```

### Step 6. Print the mean alcohol consumption per continent for every column

```python
alcohol_consumption_mean = drinks.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].mean()
display(alcohol_consumption_mean)
```

### Step 7. Print the median alcohol consumption per continent for every column

```python
alcohol_consumption_median = drinks.groupby('continent')[['beer_servings', 'spirit_servings', 'wine_servings']].median()
display(alcohol_consumption_median)
```

### Step 8. Print the mean, min and max values for spirit consumption.
#### This time output a DataFrame

```python
pirit_consumption_stats = drinks.groupby('continent')['spirit_servings'].agg(['mean', 'min', 'max'])
display(spirit_consumption_stats)
```
