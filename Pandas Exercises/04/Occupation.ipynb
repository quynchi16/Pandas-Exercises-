### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/u.user).

### Step 3. Assign it to a variable called users.

```python
users = pd.read_csv('https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/u.user', sep='|')
display(users.head())
```

### Step 4. Discover what is the mean age per occupation

```python
mean_age_per_occupation = users.groupby('occupation')['age'].mean()
display(mean_age_per_occupation)
```

### Step 5. Discover the Male ratio per occupation and sort it from the most to the least

```python
gender_counts = users.groupby(['occupation', 'gender']).size().unstack(fill_value=0)
male_ratio = (gender_counts['M'] / (gender_counts['M'] + gender_counts['F'])).sort_values(ascending=False)
display(male_ratio)
```

### Step 6. For each occupation, calculate the minimum and maximum ages

```python
min_max_ages = users.groupby('occupation')['age'].agg(['min', 'max'])
display(min_max_ages)
```

### Step 7. For each combination of occupation and gender, calculate the mean age

```python
mean_age_occupation_gender = users.groupby(['occupation', 'gender'])['age'].mean()
display(mean_age_occupation_gender)
```

### Step 8.  For each occupation present the percentage of women and men

```python
total_per_occupation = gender_counts.sum(axis=1)
percentage_gender_per_occupation = gender_counts.div(total_per_occupation, axis=0) * 100
display(percentage_gender_per_occupation)
```
