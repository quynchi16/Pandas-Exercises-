```python
url = "https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/student-mat.csv"
df = pd.read_csv(url)
display(df.head())
```

### Step 4. For the purpose of this exercise slice the dataframe from 'school' until the 'guardian' column

```python
df_sliced = df.loc[:, 'school':'guardian']
display(df_sliced.head())
```

### Step 5. Create a lambda function that will capitalize strings.

```python
capitalize_string = lambda x: x.capitalize()
```

### Step 6. Capitalize both Mjob and Fjob

```python
df_sliced['Mjob'] = df_sliced['Mjob'].apply(capitalize_string)
df_sliced['Fjob'] = df_sliced['Fjob'].apply(capitalize_string)
display(df_sliced.head())
```

### Step 7. Print the last elements of the data set.

```python
display(df_sliced.tail())
```

### Step 8. Did you notice the original dataframe is still lowercase? Why is that? Fix it and capitalize Mjob and Fjob.

```python
df['Mjob'] = df['Mjob'].apply(capitalize_string)
df['Fjob'] = df['Fjob'].apply(capitalize_string)
display(df.head())
```

### Step 9. Create a function called majority that returns a boolean value to a new column called legal_drinker (Consider majority as older than 17 years old)

```python
def majority(age):
    return age > 17

df['legal_drinker'] = df['age'].apply(majority)
display(df.head())
```

```python
for col in df.columns:
    if df[col].dtype == 'int64' or df[col].dtype == 'float64':
        df[col] = df[col] * 10

display(df.head())
```
