### Step 1. Import the necessary libraries:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/tips.csv).

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv('https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/tips.csv')
display(tips.head())
```

### Step 4. Delete the Unnamed 0 column

```python
if 'Unnamed: 0' in tips.columns:
    tips = tips.drop('Unnamed: 0', axis=1)
display(tips.head())
```

### Step 5. Plot the total_bill column histogram

```python
plt.figure(figsize=(10, 6))
sns.histplot(tips['total_bill'], kde=True, bins=15)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('Total Bill vs. Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.relplot(x='total_bill', y='tip', col='time', hue='smoker', size='size', data=tips)
plt.suptitle('Relationship between Total Bill, Tip, Size, Time, and Smoker', y=1.02)
plt.show()
```

### Step 8. Present the relationship between days and total_bill value

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='day', y='total_bill', data=tips, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Total Bill Distribution by Day')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='tip', y='day', hue='sex', data=tips, s=100, alpha=0.7)
plt.title('Tip vs. Day, differentiated by Sex')
plt.xlabel('Tip ($)')
plt.ylabel('Day of the Week')
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='day', y='total_bill', hue='time', data=tips, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Total Bill per Day, differentiated by Time (Dinner or Lunch)')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.legend(title='Time')
plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
g = sns.FacetGrid(tips, col='time', height=6)
g.map(sns.histplot, 'tip', kde=True, bins=15)
g.set_axis_labels('Tip ($)', 'Frequency')
g.set_titles('Tip Distribution for {col_name}')
plt.suptitle('Tip Value Histograms for Dinner and Lunch', y=1.02)
plt.tight_layout()
plt.show()
```

### Step 12.5. Present the number of people per day, differentiating the time (Dinner or Lunch).

```python
plt.figure(figsize=(12, 7))
sns.barplot(x='day', y='size', hue='time', data=tips, estimator=sum, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Number of People per Day, differentiated by Time (Dinner or Lunch)')
plt.xlabel('Day of the Week')
plt.ylabel('Total Number of People')
plt.legend(title='Time')
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
sns.relplot(x='total_bill', y='tip', col='sex', hue='smoker', data=tips, kind='scatter', height=6, aspect=1.2)
plt.suptitle('Total Bill vs. Tip by Sex and Smoker Status', y=1.02)
plt.show()
```

### BONUS: Create your own question and answer it using a graph.

```python
tips['tip_percentage'] = (tips['tip'] / tips['total_bill']) * 100
plt.figure(figsize=(10, 6))
sns.barplot(x='day', y='tip_percentage', hue='sex', data=tips, estimator=lambda x: sum(x)/len(x), errorbar=None, order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Average Tip Percentage by Day and Sex')
plt.xlabel('Day of the Week')
plt.ylabel('Average Tip Percentage (%)')
plt.legend(title='Sex')
plt.show()
```
